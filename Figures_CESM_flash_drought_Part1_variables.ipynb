```python

# Assuming you have two DataArrays: var1 and var3 with the same dimensions (e.g., lat, lon)
# var1: first variable DataArray
# var3: second variable DataArray

def spatial_correlation(var1, var3):
    # Flatten the spatial dimensions (lat, lon) to 1D arrays
    var1_flat = var1.values.flatten()
    var3_flat = var3.values.flatten()

    # Mask out any NaN values
    valid_mask = np.isfinite(var1_flat) & np.isfinite(var3_flat)
    var1_flat = var1_flat[valid_mask]
    var3_flat = var3_flat[valid_mask]

    # Compute the Pearson correlation coefficient
    correlation = np.corrcoef(var1_flat, var3_flat)[0, 1]

    return correlation

correlation_value_RP_PP = [0,0,0,0,0]
correlation_value_BP_PP = [0,0,0,0,0]

# Example usage:
for i in range(5): # iterating through variable list created above
    correlation_value_RP_PP[i] = spatial_correlation(variables_RP[i], variables_PP[i])
    correlation_value_RP_PP[i]

    correlation_value_BP_PP[i] = spatial_correlation(variables_BP[i], variables_PP[i])
    correlation_value_BP_PP[i]

```

```python
from scipy.stats import linregress

# Create a figure and a 2x2 grid of subplots
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# Plot in each subplot and set individual titles
titles = ['(a) Tropics (0\u00B0-10\u00B0N)', '(b) Subtropics (10\u00B0N-30\u00B0N)', '(c) Mid-latitudes (30\u00B0N-60\u00B0N)', '(d) High-latitudes (60\u00B0N-90\u00B0N)']  # Customize your titles here

# Data sets
datasets = [
    (transpiration_trop, soil_moisture_trop),
    (transpiration_subtrop, soil_moisture_subtrop),
    (transpiration_mid, soil_moisture_mid),
    (transpiration_high, soil_moisture_high)
]

for i in range(2):
    for j in range(2):
        index = i * 2 + j
        transpiration, soil_moisture = datasets[index]
        
        # Filter out NaN values
        valid_indices = ~np.isnan(transpiration) & ~np.isnan(soil_moisture)
        transpiration = transpiration[valid_indices]
        soil_moisture = soil_moisture[valid_indices]
        
        axs[i, j].scatter(transpiration, soil_moisture, alpha=0.5)
        axs[i, j].axvline(x=0, color='k', linestyle='--')  # Add vertical zero line
        axs[i, j].axhline(y=0, color='k', linestyle='--')  # Add vertical zero line
        
        # Calculate linear regression
        if len(transpiration) > 1 and len(soil_moisture) > 1:
            slope, intercept, r_value, p_value, std_err = linregress(transpiration, soil_moisture)
            regression_line = slope * transpiration + intercept
        
            # Plot regression line
            axs[i, j].plot(transpiration, regression_line, color='red', linestyle='-')

            # Annotate R-squared value
            axs[i, j].annotate(f'R² = {r_value**2:.2f}', xy=(0.05, 0.9), xycoords='axes fraction', fontsize=14, color='red')
        else:
            axs[i, j].annotate('Not enough data', xy=(0.05, 0.9), xycoords='axes fraction', fontsize=10, color='red')

        if j == 0:
            axs[i, j].set_ylabel('')
        else:
            axs[i, j].set_yticklabels([])
        if i == 1:
            axs[i, j].set_xlabel('')
        else:
            axs[i, j].set_xlabel('')
        axs[i, j].set_title(titles[i * 2 + j])  # Set individual titles
        axs[i, j].grid(True)

# Set a big title for the whole figure
fig.suptitle('Changes in mean Apr-Oct transpiration vs. top-10 cm soil moisture from CESM2 (VegCO2 - PI)', ha='center', fontsize=16, fontweight='bold')

# Set a single x-axis label in the center of the bottom
fig.text(0.55, 0.04, 'Transpiration change [mm s-1]', ha='center', va='center', fontsize=12)

# Set a single y-axis label in the center of the left side
fig.text(0.04, 0.5, 'Soil Moisture change [kg m-2]', ha='center', va='center', rotation='vertical', fontsize=12)

# Adjust layout to prevent overlap
plt.tight_layout(rect=[0.05, 0.05, 1, 0.95])

# Display the plots
plt.show()
```
